```python
class Animal:
  def __init__(self, nome, cor, especie):
    self.nome = nome
    self.cor = cor
    self.especie = especie

  def apresentar(self):
    print(f'Eu sou o {self.especie} chamado {self.nome}')


class Gato(Animal):

  def emitir_som(self):
    print('Miau!')

class Cachorro(Animal):
  pass ## não vou criar nada

class Elefante(Animal):
  pass

gato1 = Gato('Felix', 'Laranja', 'Siamese')
gato1.apresentar()
gato1.emitir_som()

cachorro1 =  Cachorro('Luke', 'Marrom', 'Pastor Gaúcho')
cachorro1.apresentar()

elefante1 = Elefante('Teddy', 'Cinza', 'Elefante Asiático')
elefante1.apresentar()
```

```python
class Pessoa:
  def __init__(self, nome, idade):
    self.nome = nome
    self.idade = idade

  def apresentar(self):
    print(f'O meu nome é {self.nome} e tenho {self.idade} anos')

class Funcionario(Pessoa):
  def __init__(self,nome, idade, cargo):
    super().__init__(nome, idade) ## vai puxar as infos da classe pai
    self.cargo = cargo

  def trabalhar(self):
    print(f'{self.nome} está trabalhando como {self.cargo}')

class Cliente(Pessoa):
  def __init__(self, nome, idade, saldo):
    super().__init__(nome, idade)
    self.saldo = saldo

  def comprar(self, valor):
    if(valor <= self.saldo):
      self.saldo -= valor
      print(f'A sua compra de {valor} foi aprovada, o seu saldo é  {self.saldo}')
    else:
      print('O valor da compra ultrapassa o seu saldo')


f1 = Funcionario('Maria', 22, 'Gerente de Contas')
f1.apresentar()
# f1.trabalhar()

c1 = Cliente('Roberto', 32, 100)
c1.apresentar()
c1.comprar(30)
```

```python
## Herança Múltipla e Herança de Multinível

class Animal():
  def __init__(self, nome):
    self.nome = nome

class Predador(Animal):
  def cacando(self):
    print(f'O animal {self.nome} está caçando!')

class Presa(Animal):
  def fugindo(self):
    print(f'O animal {self.nome} está sendo caçado!')

class Coelho(Presa):
  pass

class Tigre(Predador):
  pass

class Golfinho(Predador, Presa): ## Tem herança múltipla
  pass

coelho1 = Coelho('Bunny')
tigre1 = Tigre('Tigre')
golfinho1 = Golfinho('Golfinho')

coelho1.fugindo()
tigre1.cacando()

```
