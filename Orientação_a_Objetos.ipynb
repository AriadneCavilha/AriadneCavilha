```python
class Casa:
  def __init__(self, cor, quartos, banheiros):
    self.cor = cor
    self.quartos = quartos
    self.banheiros = banheiros

  def mostrar_cor(self):
    print(f'A cor da casa é: {self.cor}')

  def mostrar_quartos(self):
    print(f'Está casa tem {self.quartos} quartos')

  def mostrar_banheiros(self):
    print(f'Está casa tem {self.banheiros} banheiros')

  def adicionar_quarto(self):
    self.quartos += 1
    print(f'Está casa tem {self.quartos} quartos')

  def pintar_casa(self, nova_cor):
    self.cor = nova_cor
    print(f'A cor da casa foi mudada para {self.cor}')

casa1 = Casa('Azul', 2, 1)
casa2 = Casa('Vermelho', 3, 4)

print('Casa 1:')
casa1.mostrar_cor()
casa1.mostrar_quartos()
casa1.mostrar_banheiros()
casa1.adicionar_quarto()
casa1.pintar_casa('Amarelo  ')


print('\nCasa 2:')
casa2.mostrar_cor()
casa2.mostrar_quartos()
casa2.mostrar_banheiros()
```

```python
class Carro:
  def __init__(self, cor, ano):
    self.cor = cor
    self.ano = ano
    self.ligado = False
    self.seta = None

  def informacoes(self):
    print(f'A cor do carro é {self.cor} e o ano do carro é {self.ano}')

  def ligar(self):
    if(self.ligado):
      print('O carro já está ligado')
    else:
      self.ligado = True
      print('O carro foi ligado')

  def desligar(self):
    if(self.ligado):
      self.ligado = False
      print('O carro foi desligado')
    else:
      print('O carro já está desligado')


  def ligarSeta(self, direcao):
    if(self.ligado):
      self.seta = direcao
      print(f'Seta ligada para a {self.seta}')
    else:
      print('O carro precisa estar ligado')


carro1 = Carro('Preto', 2021)
carro1.informacoes()
carro1.ligar()
carro1.ligarSeta('Direita')

```
