```python
## Calcular quantos dias um produto duraria se a pessoa usar X porções por dia
quantidade_produto = int(input("Digite a quantidade do seu produto: "))
quantidade_diaria = int(input("Qual a porção que você vai usar por dia: "))

dias = quantidade_produto / quantidade_diaria
print(f'O seu produto vai durar cerca de {dias:.0f} dias')
```
